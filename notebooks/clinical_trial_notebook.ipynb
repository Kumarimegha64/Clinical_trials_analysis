```python
df = pd.read_sql(
    "SELECT * FROM clean_trials",
    engine
)
```

```python
df['age_category'].isnull().sum()
```

```python
df['age_category'] = df['age_category'].fillna(df['age'])
```

```python
df['age_category'].isnull().sum()
```

```python
(df['study_documents'].isna().mean() * 100)
```

```python
df=df.drop(columns=["study_documents"])
```

```python
df.isnull().sum()
```

```python
df['country'] = df['country'].fillna("Others")
```

```python
df['country'].value_counts()
```

```python
df=df.dropna(subset=["study_designs"])
df=df.dropna(subset=["other_ids"])
df=df.dropna(subset=["primary_completion_date_clean"])
df=df.dropna(subset=["completion_date_clean"])
```

```python
df.isnull().sum()
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
df.to_sql(
    "clean_trials",
    engine,
    if_exists="replace",
    index=False
)
```
